```python
y=ds1['Consumption']
x=ds1.drop(['Consumption','Date','DayOfWeek'],axis=1)
print(x.shape)
print(y.shape)
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2,random_state=30)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
print("Premières prédictions :", y_pred[:5])
print("Valeurs réelles :", y_test.iloc[:5].values)
```

```python
plt.scatter(y_test,y_pred)
```

```python
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=mse**0.5
r2=r2_score(y_test,y_pred)
print("MAE :",mae)
print("MSE :",mse)
print("RMSE :",rmse)
print("R2 :",r2)
```

```python
plt.scatter(y_test,y_pred, alpha=0.5)
plt.plot([y_test.min(),y_test.max()],[y_test.min(),y_test.max()],"r--")
plt.grid()
```

```python
rf=RandomForestRegressor(n_estimators=100,random_state=30)
rf.fit(x_train,y_train)
y_predrf=rf.predict(x_test)
```

```python
mae_rf=mean_absolute_error(y_test,y_predrf)
mse_rf=mean_squared_error(y_test,y_predrf)
rmse_rf=mse_rf**0.5
r2_rf=r2_score(y_test,y_predrf)
print("MAE :",mae_rf)
print("MSE :",mse_rf)
print("RMSE :",rmse_rf)
print("R2 :",r2_rf)
```

```python
plt.scatter(y_test,y_predrf, alpha=0.5)
plt.plot([y_test.min(),y_test.max()],[y_test.min(),y_test.max()],"r--")
plt.grid()
```

```python
y_train_pred = rf.predict(x_train)
r2_train = r2_score(y_train, y_train_pred)
r2_test = r2_score(y_test, y_predrf)
print("R² entraînement :", r2_train)
print("R² test :", r2_test)
```

```python
importance=pd.Series(rf.feature_importances_,index=x_train.columns)
importance=importance.sort_values(ascending=False)
print(importance)
```

```python
importance.plot(kind="bar", figsize=(10, 5))
plt.title("Importance des variables dans Random Forest")
plt.xlabel("Variables")
plt.ylabel("Importance")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.show()
```

```python
from sklearn.inspection import permutation_importance

result = permutation_importance(
    rf,
    x_test,
    y_test,
    n_repeats=10,
    random_state=30,
    scoring="r2"
)

importance_perm = pd.Series(result.importances_mean, index=x_test.columns).sort_values(ascending=True)
importance_perm.plot(kind="barh", figsize=(8, 5))
```

```python
print(importance_perm.sort_values(ascending=False))
```

```python
rf2 = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_leaf=5,
    random_state=30
)

rf2.fit(x_train, y_train)

y_train_pred2 = rf2.predict(x_train)
y_test_pred2 = rf2.predict(x_test)

print("R² entraînement :", r2_score(y_train, y_train_pred2))
print("R² test :", r2_score(y_test, y_test_pred2))

print("MAE test :", mean_absolute_error(y_test, y_test_pred2))
```

```python
scores = cross_val_score(
    rf2,
    x_train,
    y_train,
    cv=5,
    scoring="r2"
)

print("Scores des 5 folds :", scores)
print("R² moyen :", scores.mean())
print("Écart-type :", scores.std())
```
